```python
import numpy as np
import pandas as pd

# NumPy array creation
arr = np.array([10, 20, 30, 40, 50])
print("Original Array:")
print(arr)

# Array operations
print("\nArray + 5:")
print(arr + 5)

print("\nArray * 2:")
print(arr * 2)

print("\nSum of array:")
print(np.sum(arr))

print("\nMean of array:")
print(np.mean(arr))

# Creating a DataFrame
data = {
    "Name": ["Anu", "Riya", "Sneha", "Priya"],
    "Age": [20, 21, 19, 22],
    "Marks": [85, 90, 78, 92]
}

df = pd.DataFrame(data)

print("\nDataFrame:")
print(df)

# Indexing
print("\nFirst row:")
print(df.iloc[0])

print("\nMarks column:")
print(df["Marks"])

# Slicing
print("\nFirst two rows:")
print(df.iloc[0:2])

print("\nFirst three rows and Name + Marks columns:")
print(df.loc[0:2, ["Name", "Marks"]])
```
